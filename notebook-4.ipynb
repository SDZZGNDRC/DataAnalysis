```python
import pyarrow.parquet as pq
import glob
from pathlib import Path

root = Path(r'D:\Project\DataAnalysis')

jsonFiles = glob.glob(str(root / Path('tmp') / Path('OKX-Books-1INCH-EUR-400-*.json')))
```

```python
jsonFiles[:10]
```

```python
jsonl_path = 'OKX-Books-1INCH-EUR-400.jsonl'
with open(jsonl_path, 'w', encoding='utf-8') as jsonl:
    ls = []
    for f in jsonFiles:
        jf = open(f, 'r', encoding='utf-8')
        ls.append(jf.readline())
    jsonl.writelines(map(lambda f: f + '\n', ls))
```

```python
from pyarrow import json
from pyarrow.json import ReadOptions

read_options = ReadOptions(block_size=50 * 1024 * 1024)
table = json.read_json(jsonl_path, read_options=read_options)
```

```python
parquet_file = r'OKX-Books-1INCH-EUR-400.parquet'
pq.write_table(table, parquet_file)
```

```python
pf = pq.ParquetFile(parquet_file)
```

```python
metadata = pf.metadata
```
