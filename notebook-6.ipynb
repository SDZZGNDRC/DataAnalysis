```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# 从 CSV 文件读取数据
# 替换 'your_file.csv' 为实际的 CSV 文件路径
data = pd.read_csv('file_processing_times.csv')

# 确保 Timestamp 相关列被解析为 datetime 格式
data['StartTimestamp'] = pd.to_datetime(data['StartTimestamp'])
data['EndTimestamp'] = pd.to_datetime(data['EndTimestamp'])

# 创建散点图和连接线
plt.figure(figsize=(10, 6))

# 遍历每一行数据，绘制两个点并连接
for index, row in data.iterrows():
    # 提取两个时间戳和对应的 ProcessingTimeMs
    x = [row['StartTimestamp'], row['EndTimestamp']]
    y = [row['ProcessingTimeMs'], row['ProcessingTimeMs']]  # 纵轴相同
    
    # 绘制散点
    plt.scatter(x, y, color='blue', alpha=0.5)
    
    # 绘制连接线
    plt.plot(x, y, color='red', linestyle='-', alpha=0.3)

# 设置图表标题和标签
plt.title('Processing Time vs Timestamp (Start and End)')
plt.xlabel('Timestamp')
plt.ylabel('Processing Time (ms)')

# 格式化 x 轴时间显示
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d %H:%M:%S'))
plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())

# 自动旋转时间标签以避免重叠
plt.gcf().autofmt_xdate()

# 显示网格
plt.grid(True)

# 显示图表
plt.show()
```
