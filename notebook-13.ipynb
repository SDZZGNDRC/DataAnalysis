```python
import argparse
from pathlib import Path
import numpy as np
import pyarrow.parquet as pq
import pyarrow as pa
import pandas as pd
from tqdm import tqdm
from pathlib import Path

from cbookcore import BookCore
```

```python
data_dir = Path(r'E:\tmp\OKX-Trades-BTC-USDT')
pfs = list(data_dir.glob("*.parquet"))
```

```python
pf = pq.ParquetFile(pfs[0])
```

```python
pf.schema_arrow
```

```python
df = pf.read_row_group(0).to_pandas()
```

```python
df.iloc[0].get('data')
```

```python
import pandas as pd
import plotly.graph_objects as go
from datetime import datetime

# Convert millisecond Unix timestamps to datetime
df['ts'] = pd.to_datetime(df['ts'], unit='ms')

# Create the plot
fig = go.Figure()

# Plot asks and bids with different colors
fig.add_trace(go.Scatter(x=df['ts'], y=df['asks'], mode='lines', name='Asks', line=dict(color='blue')))
fig.add_trace(go.Scatter(x=df['ts'], y=df['bids'], mode='lines', name='Bids', line=dict(color='red')))

# Update layout
fig.update_layout(
    title='Asks and Bids Over Time',
    xaxis_title='Time',
    yaxis_title='Value',
    xaxis=dict(
        tickformat='%y-%m-%d %H:%M:%S',
        tickangle=45,
        nticks=10
    ),
    showlegend=True,
    width=1200,
    height=600
)

# Show the plot
fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go
from datetime import datetime

# Convert millisecond Unix timestamps to datetime
df['ts_diff'] = df['ts'].diff()
df['ts'] = pd.to_datetime(df['ts'], unit='ms')

# Create the plot
fig = go.Figure()

# Plot asks and bids with different colors
fig.add_trace(go.Scatter(x=df['ts'], y=df['ts_diff'], mode='lines', name='time diff', line=dict(color='blue')))

# Update layout
fig.update_layout(
    title='Time difference Over Time',
    xaxis_title='Time',
    yaxis_title='difference (ms)',
    xaxis=dict(
        tickformat='%y-%m-%d %H:%M:%S',
        tickangle=45,
        nticks=10
    ),
    showlegend=True,
    width=1200,
    height=600
)

# Show the plot
fig.show()
```

```python
import plotly.express as px

# 使用Plotly Express绘制直方图
fig = px.histogram(df, x='ts_diff', nbins=30, title='ts_diff的分布',
                   labels={'ts_diff': '数值', 'count': '频次'},
                   template='plotly_white')

# 自定义图形样式
fig.update_layout(
    xaxis_title="数值",
    yaxis_title="频次",
    bargap=0.1,  # 柱子间距
    showlegend=False
)

# 显示图形
fig.show()
```
