```python
from datetime import datetime, timedelta
import os

def enumerate_dates(start_str, end_str):
    """
    生成从开始到结束日期的所有日期列表，包括开始和结束日期。
    
    参数:
    start_str (str): 开始日期，格式为'YYYY-MM-DD'
    end_str (str): 结束日期，格式为'YYYY-MM-DD'
    
    返回:
    list: 包含所有日期的列表,元素为datetime.date对象
    """
    # 将字符串转换为日期对象
    start_date = datetime.strptime(start_str, '%Y-%m-%d').date()
    end_date = datetime.strptime(end_str, '%Y-%m-%d').date()
    
    dates = []
    current_date = start_date
    
    if start_date <= end_date:
        # 正向生成日期
        delta = timedelta(days=1)
        while current_date <= end_date:
            dates.append(current_date)
            current_date += delta
    else:
        # 反向生成日期
        delta = timedelta(days=-1)
        while current_date >= end_date:
            dates.append(current_date)
            current_date += delta
    
    return dates

start_date = '2025-01-13'
end_date = '2025-02-16'
dates = list(map(lambda d: d.strftime('%Y-%m-%d'), enumerate_dates(start_date, end_date)))

dates_files = {}

for date in dates:
    dates_files[date] = os.listdir(rf"E:\datapool\{date}")

dates_splitted = {}

for date in dates:
    dates_splitted[date] = list(map(lambda f: f.split("-"), dates_files[date]))

```
